# 💖 Breakup Buddy
**An AI-Powered Self-Love Journal for Breakups & Divorce Recovery**

> Whether you're grieving a love lost or just feeling emotionally overwhelmed, this app is here to help you process, grow, and heal.

# Why This Matters

> Breakups and heartbreak can be quietly devastating. Not everyone has a friend they can talk to. This app is a private, judgment-free space to express, reflect, and heal, with the help of compassionate generative AI.

# What This App Does

This app is designed to help you:


> Reflect daily through journaling
> 
> Receive compassionate responses powered by Gemini LLM
> 
> Generate healing affirmations, closure letters, and loving reflections
> 
> Track your emotional healing over time
> 
> Retrieve past thoughts that match your current mood


# Features
> Gradio UI
> 
> Gemini LLM reflections
> 
> Daily affirmations
> 
> Closure letter generator
> 
> Journal saved to CSV
> 
> Healing streak & mood chart
> 
> Chroma vector search for past reflections
> 
> Poetic cover & gentle UX



# Generative AI Capabilities Used


> **Few-shot prompting**: Guiding Gemini to act like an empathetic friend
>
> **Embeddings + ChromaDB**: Store & retrieve journal entries based on similarity
>
> **Controlled generation**: Generate affirmations and poetic letters
>
> **RAG-style retrieval**: Matching current thoughts to past reflections


# Tech Stack



> * Gradio: For the interactive web interface
> * Gemini Pro (via Google Generative AI API): To generate emotionally intelligent reflections
> * Chroma DB: To store and retrieve past reflections with similarity search
> * CSV (Pandas): For journaling history and tracking progress


# How It Works

**Journaling**

> Users write freely about their feelings.
> Gemini reflects back with a kind, encouraging response tailored to their mood.


**Logging**

> Each entry (date, mood, text) is saved to a CSV.The app tracks how often you show up to write.


**Mood Tracker**
> Your emotional streak is charted across time — showing how often you journal.

**Smart Recall**
> Using Chroma vector DB, the app finds similar past reflections based on today’s writing — helping you notice patterns or growth.

**Extras**

> Self-love affirmationsHealing quotes
> Closure letter generator

```python
# === Imports ===
import os  # Provides a way of using operating system-dependent functionality like reading or writing to the file system
import gradio as gr  # Library for building UIs in Python
import google.generativeai as genai  # Interface for Google's generative AI services
import pandas as pd  # Data manipulation library, used to handle CSV files
import datetime  # Provides date and time manipulation utilities
import random  # Provides functions for generating random numbers and selections
import chromadb  # Used for storing and querying embeddings in a vector database
import matplotlib.pyplot as plt  # Library for plotting data
from io import BytesIO  # Handles byte data streams, useful for saving and sending images in memory
from PIL import Image  # Image processing library to handle images
from sentence_transformers import SentenceTransformer  # Sentence embedding model for text similarity
from kaggle_secrets import UserSecretsClient  # Secure management of API keys


# === Constants & Paths ===
WORKING_DIR = "/kaggle/working"  # Set working directory to Kaggle's working directory
CSV_FILE = os.path.join(WORKING_DIR, "journal_entries.csv")  # Path to the CSV file where journal entries will be stored
CHROMA_DB_DIR = os.path.join(WORKING_DIR, "chroma_db")  # Directory where ChromaDB vector data will be stored


# === Gemini API Setup ===
user_secrets = UserSecretsClient()  # Client to retrieve secrets securely from Kaggle
GEMINI_API_KEY = user_secrets.get_secret("GEMINI_API_KEY")  # Get the Gemini API key from secrets
genai.configure(api_key=GEMINI_API_KEY)  # Set the Gemini API key to access the generative AI model
model = genai.GenerativeModel(model_name="models/gemini-1.5-pro-latest")  # Load the specific model for generating reflections


# === Embedding Model Setup ===
embedder = SentenceTransformer("all-MiniLM-L6-v2")  # Load a pre-trained SentenceTransformer model for embedding text


# === Chroma Vector DB Setup ===
chroma_client = chromadb.PersistentClient(path=CHROMA_DB_DIR)  # Create a persistent client for ChromaDB
collection = chroma_client.get_or_create_collection(name="journal_entries")  # Create or retrieve the collection where journal entries will be stored


# === Create CSV File if Not Exists ===
if not os.path.exists(CSV_FILE):  # Check if the CSV file exists
    # If it doesn't exist, create a new CSV with the appropriate columns
    pd.DataFrame(columns=["date", "mood", "entry"]).to_csv(CSV_FILE, index=False)


# === Mood Options ===
mood_options = ["😢 Sad", "😡 Angry", "😔 Lonely", "💭 Nostalgic", "😐 Numb", "🌤️ Hopeful"]  # List of mood options for the user to select


# === Core Functionality: Reflect on Journal Entry ===
def reflect_journal(mood, entry):
    """
    Generate a kind reflection using Gemini and store entry in CSV and ChromaDB.
    
    Parameters:
        mood (str): The current mood of the user.
        entry (str): The user's journal entry.
    
    Returns:
        str: Reflection generated by the Gemini model.
    """
    # If no entry is provided, ask the user to write something
    if not entry.strip():
        return "Please write something from your heart 📝"
    
    # Generate the prompt for the Gemini model based on mood and journal entry
    prompt = f"""
    You are a kind friend helping someone recover from heartbreak.
    They feel {mood} today and wrote:
    \"\"\"{entry}\"\"\"
    
    Reflect back gently with kindness, warmth. Offer reassurance and one tiny self-kindness idea.
    """
    
    try:
        # Call Gemini to generate a reflection based on the prompt
        response = model.generate_content(prompt).text
    except Exception as e:
        # Handle any error from Gemini and return the error message
        return f"Error from Gemini: {str(e)}"

    # Save the journal entry with the current date, mood, and text to CSV
    today = datetime.date.today().isoformat()  # Get today's date
    df = pd.DataFrame([[today, mood, entry]], columns=["date", "mood", "entry"])  # Prepare the data as a DataFrame
    df.to_csv(CSV_FILE, mode='a', index=False, header=False)  # Append the entry to the CSV file

    # Store the journal entry in ChromaDB as an embedding
    embedding = embedder.encode(entry).tolist()  # Get the embedding of the journal entry
    unique_id = f"{today}_{len(df)}"  # Create a unique ID based on the date and entry count
    collection.add(documents=[entry], embeddings=[embedding], ids=[unique_id])  # Store the entry and embedding in ChromaDB

    return response  # Return the reflection generated by Gemini

    
# === Core Functionality: Retrieve Similar Past Reflections ===
def retrieve_similar_entries(current_entry):
    """
    Retrieve similar past journal entries from ChromaDB based on the current entry.
    
    Parameters:
        current_entry (str): The user's current journal entry.
    
    Returns:
        str: Similar past entries from ChromaDB or a message if no similar entries are found.
    """
    # If no entry is provided, ask the user to write something
    if not current_entry.strip():
        return "Write something first to find similar past reflections 📝"
    
    try:
        # Get the embedding of the current entry
        query_embedding = embedder.encode(current_entry).tolist()
        # Query ChromaDB to retrieve the 3 most similar entries
        results = collection.query(query_embeddings=[query_embedding], n_results=3)
        similar = results.get("documents", [[]])[0]  # Retrieve the similar documents
        # Return the similar entries or a message if no entries are found
        # Update to join the multiple past reflections correctly
        if similar:
            return "\n\n---\n\n".join(similar)  # Join the similar documents with a separator
        else:
            return "No similar past reflections found."
    except Exception as e:
        # Handle any error while querying ChromaDB
        return f"Error retrieving: {str(e)}"


# === Core Functionality: Generate Random Affirmation ===
def random_affirmation():
    """
    Return a randomly selected affirmation for emotional support.
    
    Returns:
        str: A randomly chosen affirmation.
    """
    # List of affirmations
    affirmations = [
        "You are not alone, My Love!, I am with you always.",
        "You are enough, My Love!, exactly as you are.",
        "You’re allowed to begin again My Love!, softly.",
        "You are healing even when it’s quiet, My Love!.",
        "You are loved, even when you don’t feel it, My Love!.",
        "You are not starting over — you are starting wiser, My Love!."
    ]
    # Return a randomly selected affirmation from the list
    return random.choice(affirmations)


# === Core Functionality: Generate a Poetic Closure Letter ===
def generate_closure():
    """
    Use Gemini to create a poetic closure letter after breakup.
    
    Returns:
        str: The poetic closure letter generated by Gemini.
    """
    # Create the prompt to generate a poetic closure letter
    prompt = """
    Write a poetic closure letter from someone healing after a breakup.
    Don't blame. Emphasize acceptance, gratitude, and moving on.
    """
    try:
        # Call Gemini to generate the closure letter based on the prompt
        return model.generate_content(prompt).text
    except Exception as e:
        # Handle any error from Gemini and return the error message
        return f"Error from Gemini: {str(e)}"


# === Core Functionality: Mood Chart Visualization ===
def plot_mood_chart():
    """
    Plot mood trend over time from CSV entries.
    
    Returns:
        PIL.Image: The mood chart image for display.
    """
    # Check if the CSV file exists
    if not os.path.exists(CSV_FILE):
        return None
    
    # Read the CSV file containing journal entries
    df = pd.read_csv(CSV_FILE)
    if df.empty:
        return None
    
    # Convert dates from string to datetime objects for plotting
    df["date"] = pd.to_datetime(df["date"])

    # Convert mood to index values to plot on the graph
    df["mood_index"] = df["mood"].apply(lambda x: mood_options.index(x))

    # Plot setup: Create a figure and axis for plotting
    fig, ax = plt.subplots(figsize=(7, 3))
    ax.plot(df["date"], df["mood_index"], marker='o', linestyle='-')  # Plot the mood index over time
    ax.set_yticks(range(len(mood_options)))  # Set the y-ticks to correspond to mood options
    ax.set_yticklabels(mood_options)  # Label the y-ticks with mood names
    ax.set_title("📈 Your Healing Mood Chart, My Love")  # Set the title
    ax.set_xlabel("Date")  # Set the x-axis label
    ax.set_ylabel("Mood")  # Set the y-axis label

    # Improve x-axis readability (rotate the date labels)
    fig.autofmt_xdate()

    # Save the plot to a BytesIO buffer and return it as a PIL image
    buf = BytesIO()
    plt.tight_layout()
    plt.savefig(buf, format="png")
    plt.close(fig)
    buf.seek(0)
    return Image.open(buf)

# === Gradio UI Setup ===
with gr.Blocks(title="💖 Breakup Buddy") as app:
    # Markdown text for the UI title and description
    gr.Markdown("<h1 style='text-align: center;'>💖 Breakup Buddy</h1>")
    gr.Markdown("<h2 style='text-align: center;'>An AI-powered journal that listens, reflects, and heals with you.</h2>")
    gr.Markdown("<p style='text-align: center;'>“My Love, your heart is a garden. Even now, something beautiful is growing.”</p>")
    
    # Custom CSS Styling for the UI
    gr.HTML("""
    <style>
    body {
        font-family: 'Inter', sans-serif;
        background: #fffafc;
        text-align: center;
        font-size: 20px;
    }
    h1, h2 {
        color: #d63384;
        font-size: 23px;
    }
    p {
        font-style: italic;
        color: #6c757d;
        font-size: 20px;
    }
    textarea, input, select {
        border-radius: 12px !important;
        border: 1px solid #ffd6e0 !important;
        background-color: #fff0f5 !important;
        box-shadow: 0 4px 8px rgba(0, 0, 0, 0.1) !important;
        padding: 20px;
        margin: 30px;
    }
    button {
        background-color: #620F99 !important;
        color: white !important;
        border-radius: 12px !important;
        padding: 12px 20px;
        box-shadow: 0 4px 8px rgba(0, 0, 0, 0.1) !important;
        border: none !important;
    }
    .gr-button:hover {
        background-color: #9a48c9 !important; 
    }
    .gr-button:active {
        background-color: #8e3cc3 !important;
    }
    .gr-markdown {
        font-family: 'Inter', sans-serif;
    }
    /* Styling the labels to be bold and large */
    .gr-label {
        font-weight: bold !important;
        font-size: 20px !important;
    }
    </style>
    """)

    # === Journal Input Section ===
    mood = gr.Dropdown(mood_options, label="How are you feeling today, my love?", value="😐 Numb")
    entry = gr.Textbox(lines=6, label="What's on your heart, my love?", placeholder="Let it all out...i want to listen")

    reflect_btn = gr.Button("U shared, now my turn, my love? (Click me)")
    output = gr.Textbox(label="AI Reflection")
    reflect_btn.click(reflect_journal, inputs=[mood, entry], outputs=output)

    # === Retrieve Similar Reflections ===
    gr.Markdown("## Past Reflections")
    retrieve_btn = gr.Button("Let's Check ur Past moods, my love! (Click me)")
    retrieved_out = gr.Textbox(label="Your Past Reflections", placeholder="Let me find what u shared when u were in similar mood" , lines=6)
    retrieve_btn.click(fn=retrieve_similar_entries, inputs=[entry], outputs=retrieved_out)

    # === Affirmation Section ===
    gr.Markdown("---")
    aff_btn = gr.Button("Would u like a little self-love, my dear? (Click me)")
    aff_out = gr.Textbox(label="Take a moment to read this, my love:")
    aff_btn.click(fn=random_affirmation, outputs=aff_out)

    # === Closure Letter Section ===
    gr.Markdown("---")
    closure_btn = gr.Button("Would u like some closure, my dear? (Click me)")
    closure_out = gr.Textbox(label="Take a moment to read this love poem I wrote for you, my love:", lines=10)
    closure_btn.click(fn=generate_closure, outputs=closure_out)

    # === Mood Tracker Chart ===
    gr.Markdown("---")
    gr.Markdown("### 📈 Mood Tracker")
    chart_btn = gr.Button("Let's Check ur Healing Progress, my love? (Click me)")
    chart_output = gr.Image(type="pil", label="Mood Over Time")
    chart_btn.click(fn=plot_mood_chart, outputs=chart_output)

# === Launch the App ===
app.launch()

```
